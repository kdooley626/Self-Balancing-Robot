# LQR Gains Derivation
## Define variables
R=wheel radius \
$\theta$ = Pendulum tilt angle from vertical \
x = Horizontal position of wheel axis \
L = Length from wheel axis to pendulum COM \
M = Pendulum mass \
m = Wheel mass \
$I_w$ = Wheel moment of inertia \
$I_{cm}$ = Pendulum moment of intertia about COM\
$\dot{\phi}$ = Wheel angular velocity \
$\tau$ = Motor torque 

## Determine Lagrangian
$$\mathcal{L}=T-V$$
### Total Kinetic Energy
#### Total Kinetic Energy of the Pendulum
$$T_p = \frac{1}{2}M\left(\mathbf{v}_{cm}\cdot\mathbf{v}_{cm}\right)+\frac{1}{2}I_{cm}\dot{\theta}^2$$
$$\mathbf{p}_{cm}=\left(x+L\sin(\theta), L\cos(\theta)\right)$$
$$\frac{d}{dt}\mathbf{p}_{cm}=\mathbf{v}_{cm}=\left(\dot{x}+L\dot{\theta}\cos(\theta),\,-L\dot{\theta}\sin(\theta)\right)$$
$$T_{p}=\frac{1}{2}M\left(\dot{x}^2+2\dot{x}\dot{\theta}L\cos(\theta)+L^2\dot{\theta}^2\right)+\frac{1}{2}I_{cm}\dot{\theta}^2$$
#### Total Kinetic Energy of the wheel
$$T_w = \frac{1}{2}I_w\dot{\phi}^2 + \frac{1}{2}m\dot{x}^2$$
> **Roll without:** $ \dot{x}=R\dot{\phi} \implies \dot{\phi} = \frac{\dot{x}}{R} $
$$T_w = \frac{1}{2}\dot{x}^2\left(\frac{I_w}{R^2}+m\right)$$
#### Total Kinetic Energy
$$T = \frac{1}{2}M\left(\dot{x}^2+2\dot{x}\dot{\theta}L\cos(\theta)+L^2\dot{\theta}^2\right)+\frac{1}{2}I_{cm}\dot{\theta}^2+\frac{1}{2}\dot{x}^2\left(\frac{I_w}{R^2}+m\right) $$
### Total Potential Energy
#### Total Potential Energy of the Pendulum
$$V_p=MgL\cos(\theta)$$
#### Total Potential Energy of the wheel
$$V_w=0$$
#### Total Potential Energy
$$V=MgL\cos(\theta)$$
### Lagrangian
$$\mathcal{L}=T-V$$
$$\mathcal{L}=\frac{1}{2}M\left(\dot{x}^2+2\dot{x}\dot{\theta}L\cos(\theta)+L^2\dot{\theta}^2\right)+\frac{1}{2}I_{cm}\dot{\theta}^2+\frac{1}{2}\dot{x}^2\left(\frac{I_w}{R^2}+m\right) - MgL\cos(\theta)$$

## Virtual Work
$$\delta W = \tau \delta\phi-\tau\delta\theta$$
$$\delta W = \tau\delta \left(\frac{x}{R}\right)-\tau\delta\theta$$
$$\delta W = \frac{\tau}{R}\delta x -\tau\delta\theta$$

## Euler Lagrange Equations
$$\frac{d}{dt}\left(\frac{\partial\mathcal{L}}{\partial \dot{q}_j}\right)-\frac{\partial\mathcal{L}}{\partial q_j} = Q_j$$
Use Euler Langrange to solve for $\ddot{x}$ and $\ddot{\theta}$
$$\frac{d}{dt}\left(\frac{\partial\mathcal{L}}{\partial \dot{x}}\right)-\frac{\partial\mathcal{L}}{\partial x} = \frac{\tau}{R}$$

$$\frac{d}{dt}\left(\frac{\partial\mathcal{L}}{\partial \dot{\theta}}\right)-\frac{\partial\mathcal{L}}{\partial \theta} = -\tau$$

> **To simplify final equation let:** $c=\frac{1}{\left( ML\right)^2-\left(M+m+\frac{I_w}{R^2}\right)\left(ML^2+I_{cm}\right)}$

$$\ddot{x}=cg(ML)^2\theta-\tau\left(ML+\frac{L^2M}{R}+\frac{I_{cm}}{R}\right)c $$
$$\ddot{\theta}=-cgML\left(M+m+\frac{I_w}{R^2}\right)\theta + \tau \left(M+m+\frac{I_w}{R^2}+\frac{ML}{R}\right)c $$

## State Space Equation
$$ \mathbf{s} = \begin{bmatrix} \theta \\ \dot{\theta} \\ x \\ \dot{x} \end{bmatrix} $$
$$ \mathbf{\dot{s}} = A \mathbf{s(t)}+B u(t) $$
A = System Matrix \
u(t) = Input vector (control signals) ($\tau$) \
$\mathbf{s(t)}$ = State vector 
### Matrices A and B Defined Below

> **To simplify final equation let:** $c=\frac{1}{\left( ML\right)^2-\left(M+m+\frac{I_w}{R^2}\right)\left(ML^2+I_{cm}\right)}$
$$A = \begin{bmatrix}
0      & 1  & 0 & 0 \\
A_{21} & 0  & 0 & 0 \\
0     & 0  & 0 & 1 \\
A_{41} & 0  & 0 & 0 \end{bmatrix}$$


$$A_{21}=-cgML\left(M+m+\frac{I_w}{R^2}\right)$$
$$A_{41}=cg\left(ML\right)^2$$
$$B = \begin{bmatrix} 0 \\ B_{2} \\ 0 \\ B_{4} \end{bmatrix}$$

$$B_{2}=\left(M+m+\frac{I_w}{R^2}+\frac{ML}{R}\right)c$$
$$B_{4}=\left(ML+\frac{L^2M}{R}+\frac{I_{cm}}{R}\right)c$$
